### Area of a circle is calculated as follows: area = π r². Write a function that calculates area_of_circle.

```python
def area_of_circle(radius):
    '''Return the area of a circle given its radius'''
    import math
    area = math.pi * radius ** 2
    print(f'The area of a circle with radius {radius} is: {area}')

area_of_circle(7)
```

### Write a function that computes the volume of a sphere given its radius.
The volume of a sphere is given as $$\frac{4}{3} πr^3$$</p>

```python
def volume_of_sphere(radius):
    '''Return the volume of a sphere given its radius'''
    import math
    volume = (4/3) * math.pi * radius ** 3
    print(f'The volume of a sphere with radius {radius} is: {volume}')

volume_of_sphere(5)
```

### Write a function which converts °C to °F
### Temperature in °C can be converted to °F using this formula: °F = (°C x 9/5) + 32.

### Write a function called check-season, it takes a month parameter and returns the season: Autumn, Winter, Spring or Summer.

```python
def check_season(month):
    '''Return the season for a given month'''

    if month in ['december', 'january', 'february']:
        season = 'Winter'
    elif month in ['march', 'april', 'may']:
        season = 'Spring'
    elif month in ['june', 'july', 'august']:
        season = 'Summer'
    elif month in ['september', 'october', 'november']:
        season = 'Autumn'
    else:
        season = 'Invalid month name'
    print(f'The season in {month} is: {season}')

check_season('september')
```

### Write a function named print_list. It takes a list as a parameter and it prints out each element of the list.

```python
def print_list(_list):

    index = 0
    while index < len(_list):
        print(f'Index {index} : {_list[index]}')
        index += 1

print_list(['apple', 'banana', 'cherry'])
```

### Write a function named sum_of_numbers. It takes two parameters min_num and max_num and it adds all the numbers in that range.

```python
def sum_of_numbers(min_num, max_num):
    '''Return the sum of all numbers in a given range'''
    _sum = 0
    for num in range (min_num , max_num + 1 ):
        _sum += num
    print(f'The sum of numvers from {min_num} to {max_num} is : {_sum}')
sum_of_numbers(3,7)
```

### Write a function named sum_of_numbers. It takes two parameters min_num and max_num and it adds all the `odd` numbers in that range.

```python
def sum_of_numbers(min_num, max_num):
    '''Return the sum of all odd numbers in a given range'''
    _sum = 0
    for num in range (min_num , max_num + 1 ):
        if num % 2 != 0:
         _sum += num
    print(f'The sum of numvers from {min_num} to {max_num} is : {_sum}')
sum_of_numbers(3,7)
```

### Write a Python function that accepts a string and calculates the number of upper case letters and lower case letters.

    Sample String : 'Hello Mr. Rogers, how are you this fine Tuesday?'
    Expected Output :
    No. of Upper case characters : 4
    No. of Lower case Characters : 33

```python
def number_of_upper_and_lower_case_letters (word) :
    cnt_upper = 0
    cnt_lower = 0
    for letter in word :
        if letter.isupper():
            cnt_upper += 1
        elif letter.islower():
            cnt_lower += 1
    print(f'{word} has {cnt_upper} upper case letters and {cnt_lower} lower case letters.')


number_of_upper_and_lower_case_letters ('Hello Mr. Rogers, how are you this fine Tuesday?')
```

### Write a Python function that checks whether a passed in string is palindrome or not.

Note: A palindrome is word, phrase, or sequence that reads the same backward as forward, e.g., `madam` or `nurses run`.

```python
def palindrome_str(word : str) :
    word = word.replace(" " , "")
    # word.strip()
    rev_word = word[::-1]
    if word == rev_word :
        print(f'{word} is a palindrome.')
    else:
        print(f'{word} is not a palindrome.')
palindrome_str('nurses run')
palindrome_str('madam')
palindrome_str('121')
palindrome_str('hello')

```
